````python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")

DATA_YAML_PATH = "./../data/data.yaml"

results = model.train(
    data=DATA_YAML_PATH,
    imgsz=36,
    epochs=100,
    batch=256*5,
    workers=0,
    cache="disk",
    patience=30,
    device=0,
)
````

The model.train call automatically uses the validation data as we have provided it in data.yaml.  


Monitor performance on the validation set during training (used for model selection and early stopping if enabled).


After training is complete, evaluate once on the test set:
```python
metrics = model.val(
    data="data.yaml",
    split="test"
)
```
This gives an unbiased estimate of how well the final model generalizes, because the test set was never used to influence training or model selection.
